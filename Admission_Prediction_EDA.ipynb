# Exploratory Data Analysis – Graduate Admissions Dataset

This project explores a graduate admissions dataset using Exploratory Data Analysis (EDA) techniques.
The goal is to understand the distribution, variability, and relationships between academic factors
such as GRE Score, TOEFL Score, CGPA, and admission chances.


## Libraries and Tools

We import the necessary Python libraries for data manipulation, numerical analysis, and visualization.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Dataset Loading

The dataset is loaded from a CSV file containing academic and profile-related features
used to predict graduate admission chances.


```python
data = pd.read_csv('Admission_Predict.csv')
```

## Initial Data Exploration

We begin by inspecting the first few rows of the dataset to understand its structure and variables.


```python
data.head()
```

The shape of the dataset helps us understand how many observations and features are available.


```python
data.shape
```

We examine data types and check for missing values to assess overall data quality.


```python
data.info()
```

## Descriptive Statistics

We calculate the mean values for numerical variables to understand central tendencies.


```python
data.mean()
```

The mode of the GRE Score highlights the most frequently occurring values.


```python
data['GRE Score'].mode()
```

The median is computed to provide a robust measure of central tendency, less sensitive to outliers.


```python
data.median()
```

Quartiles are calculated to analyze data dispersion and identify the spread of values.


```python
data.quantile(q=0.25)
```

```python
data.quantile(q=0.50)
```

```python
data.quantile(q=0.75)
```

We compute mean, median, and mode specifically for the GRE Score to compare different
measures of central tendency.
The computed values are printed for quick reference and comparison.


```python
mean = data['GRE Score'].mean()
median = data['GRE Score'].median()
mode = data['GRE Score'].mode()

print('Mean: ', mean, '\nMedian: ', median, '\nMode: ', mode[0])
```

## Distribution of GRE Scores

A histogram is used to visualize the distribution of GRE scores.
Vertical lines represent the mean, median, and mode to compare their positions.


```python
plt.Figure(figsize=(10,5))
plt.hist(data['GRE Score'], bins=100, color='lightblue')
plt.axvline(mean, color='green', label='Mean')
plt.axvline(median, color='blue', label='Median')
plt.axvline(mode[0], color='red', label='Mode1')
plt.axvline(mode[1], color='red', label='Mode2')

plt.xlabel('GRE Score')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

**Observation:**  
The GRE Score distribution appears approximately normal, with slight deviations,
indicating a relatively balanced dataset.


The `describe()` method provides a consolidated statistical summary of the dataset.

```python
data.describe()
```

## Interquartile Range (IQR)

The IQR is calculated for the TOEFL Score to measure variability and detect potential outliers.

```python
data['TOEFL Score'].quantile(0.75) - data['TOEFL Score'].quantile(0.25)
```

A boxplot is used to visually represent the distribution, quartiles, and potential outliers
of the TOEFL Score.

```python
data.boxplot(column='TOEFL Score', return_type='axes', figsize=(8,8))

plt.text(x=0.74, y=112, s='3rd Quartile')
plt.text(x=0.8, y=107, s='Median')
plt.text(x=0.75, y=103, s='1st Quartile')
plt.text(x=0.9, y=92, s='Min')
plt.text(x=0.9, y=120, s='Max')
plt.text(x=0.7, y=107.5, s='IQR', rotation=90, size=25)
```

## Skewness Analysis

## Normality Check – GRE Score

The GRE Score distribution is compared with a fitted normal distribution
to evaluate how closely it follows a Gaussian pattern.

```python
# HISTOGRAM FOR CHECKING SKEWNESS
import scipy.stats as stats

# Convert pandas DataFrame object to Numpy array and sort
h = np.asarray(data['GRE Score'])
h = sorted(h)

# Use the scipy stats model to fit a normal distribution with same mean and standar deviation
fit = stats.norm.pdf(h, np.mean(h), np.std(h))

# Plot both series in a histogram
plt.plot(h, fit, '-', linewidth=2, label='Normal distribution with same mean and var')
plt.hist(h, density=True, bins=100, label='Actual distribution')
plt.legend()
plt.show()
```

**Conclusion:**  
The GRE Score shows behavior close to a normal distribution, validating
the use of parametric statistical methods.
